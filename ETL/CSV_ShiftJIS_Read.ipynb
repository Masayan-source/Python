```python
import pandas as pd
```

```python
# ファイルパス
file_path = r"C:\Users\藤井\Documents\Python\Data\StudentScore_UTF-8.csv"

# Excel 読み込み（Sheet1）
df = pd.read_csv(file_path, encoding="utf-8")
```

```python
# 型変換（Power Query の TransformColumnTypes 相当）
df = df.astype({
    "学籍番号": "int64",
    "テスト点数": "int64",
    "出席率": "float",
    "クラス": "string"
})
df
```

```python
# 日付列の変換
df["受験日"] = pd.to_datetime(df["受験日"])
```

```python
# 確認
print(df)
print(df.dtypes)
```
