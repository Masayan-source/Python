```python
import pandas as pd
import openpyxl
```

```python
# ファイルパス
file_path = r"C:\Users\藤井\Documents\Python\Data\StudentScore.xlsx"

# Excel 読み込み（Sheet1）
df = pd.read_excel(file_path, sheet_name="Sheet1")
```

```python
# 日付列の変換
df["受験日"] = pd.to_datetime(df["受験日"])
```
